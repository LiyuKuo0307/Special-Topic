```python
import re
import unicodedata

import jieba
import pandas as pd
from opencc import OpenCC


class FakeNewsAnalyzer:

    def __init__(self):

        self.df = None

        self.cc = OpenCC("t2s")

        self.stopwords = {
            "的","了","是","在","有","與","及","和","就",
            "而","被","把","讓","並","呢","嗎","啊",
            "一個","我們","你們","他們","可以","如果",
            "就是","這個","那個","沒有","不是","因為"
        }
        self.url_pattern = re.compile(r"https?://\S+|www\.\S+")
        self.email_pattern = re.compile(r"\S+@\S+")
        self.html_pattern = re.compile(r"<.*?>")
        self.mention_pattern = re.compile(r"@\w+")

    # =====================================================
    # Load CSV
    # =====================================================

    def load_csv(self, csv_path):

        self.df = pd.read_csv(
            csv_path,
            encoding="utf-8",
            dtype={"id": str}
        )

        required_columns = [
            "id",
            "platform",
            "title",
            "content",
            "time",
            "like_count",
            "comment_count",
            "url"
        ]

        missing = [
            col
            for col in required_columns
            if col not in self.df.columns
        ]

        if len(missing) > 0:

            raise ValueError(
                f"CSV缺少欄位：{missing}"
            )

        # ---------------------------
        # 基本欄位整理
        # ---------------------------

        self.df["content"] = (
            self.df["content"]
            .fillna("")
            .astype(str)
        )

        self.df["title"] = (
            self.df["title"]
            .fillna("")
            .astype(str)
        )

        self.df["platform"] = (
            self.df["platform"]
            .fillna("Unknown")
            .astype(str)
        )

        self.df["url"] = (
            self.df["url"]
            .fillna("")
            .astype(str)
        )

        # content 必填
        self.df = self.df[
            self.df["content"].str.strip() != ""
        ]

        # 日期

        self.df["time"] = pd.to_datetime(
            self.df["time"],
            errors="coerce"
        )

        self.df = self.df.dropna(
            subset=["time"]
        )

        self.df["date"] = self.df["time"].dt.strftime(
            "%Y-%m-%d"
        )

        # 數值

        self.df["like_count"] = pd.to_numeric(
            self.df["like_count"],
            errors="coerce"
        ).fillna(0).astype(int)

        self.df["comment_count"] = pd.to_numeric(
            self.df["comment_count"],
            errors="coerce"
        ).fillna(0).astype(int)

        # ---------------------------
        # 新增分析欄位
        # ---------------------------

        self.df["full_text"] = (
            self.df["title"] + " " + self.df["content"]
        )

        self.df["clean_content"] = ""

        self.df["clean_title"] = ""

        self.df["text_length"] = 0

        self.df["tokens"] = None

        self.df["keywords"] = None

        self.df["emotion"] = None

        self.df["emotion_score"] = None

        self.df["is_anomaly"] = False

        self.df = self.df.reset_index(drop=True)

        print("[INFO] CSV Loaded Successfully")

    # =====================================================
    # Dataset Information
    # =====================================================

    def info(self):

        print("=" * 60)

        print("Fake News Analyzer V2")

        print("=" * 60)

        print(f"Posts          : {len(self.df)}")

        print(
            f"Platforms      : {self.df['platform'].nunique()}"
        )

        print(
            f"Date Range     : {self.df['date'].min()} ~ {self.df['date'].max()}"
        )

        print()

        print("Platform Distribution\n")

        counts = self.df["platform"].value_counts()

        for platform, count in counts.items():
           print(f"{platform:<10} : {count}")

        print()

        print(
            f"Missing Content : {(self.df['content']=='').sum()}"
        )

        duplicate_count = self.df.duplicated(
          subset=["platform", "title", "content", "time"]
        ).sum()
        print(f"Duplicate Posts : {duplicate_count}")
        print("=" * 60)

    # =====================================================
    # Preview
    # =====================================================

    def preview(self, n=5):

        columns = [
        "id",
        "platform",
        "title",
        "clean_title",
        "content",
        "clean_content",
        "text_length"
        ]

        print(self.df[columns].head(n))
    # =====================================================
    # Clean Text
    # =====================================================

    def clean_text(self, text):

       if pd.isna(text):
         return ""

       text = str(text)

       # 全形轉半形
       text = unicodedata.normalize("NFKC", text)

       # URL
       text = self.url_pattern.sub(" ", text)

       # Email
       text = self.email_pattern.sub(" ", text)

       # HTML
       text = self.html_pattern.sub(" ", text)

       # @帳號
       text = self.mention_pattern.sub(" ", text)

       # hashtag
       text = text.replace("#", " ")

       # Emoji（大部分）
       text = re.sub(
           "["
           "\U00010000-\U0010ffff"
           "]",
           " ",
           text
       )

       # 保留：
       # 中文
       # 英文
       # 數字
       text = re.sub(
           r"[^A-Za-z0-9\u4e00-\u9fff\s]",
           " ",
           text
       )

       # 多空白
       text = re.sub(r"\s+", " ", text)

       return text.strip()

    # =====================================================
    # Clean Data
    # =====================================================

    def clean_data(self):

       print("[INFO] Cleaning Data...")

       self.df["clean_content"] = self.df["content"].apply(
           self.clean_text
       )

       self.df["clean_title"] = self.df["title"].apply(
           self.clean_text
       )

       self.df["full_text"] = (
           self.df["clean_title"] +
           " " +
           self.df["clean_content"]
       ).str.strip()

       self.df["text_length"] = (
           self.df["clean_content"]
           .str.len()
       )

       print("[INFO] Data Cleaning Finished")
```

```python
# ==========================================
# Load Custom Word Dictionary
# ==========================================

import pandas as pd
import jieba

def load_custom_words(path="custom_words.csv"):

    # 嘗試使用 UTF-8 讀取
    try:
        df_words = pd.read_csv(
            path,
            encoding="utf-8-sig"
        )
    except UnicodeDecodeError:
        print("[WARNING] UTF-8 讀取失敗，改用 Big5")
        df_words = pd.read_csv(
            path,
            encoding="big5"
        )

    # 檢查欄位
    if "word" not in df_words.columns:
        raise ValueError(
            f"找不到 word 欄位，目前欄位為：{df_words.columns.tolist()}"
        )

    words = (
        df_words["word"]
        .dropna()
        .astype(str)
        .str.strip()
    )

    custom_words = []

    for word in words:

        if word == "":
            continue

        jieba.add_word(word)
        custom_words.append(word)

    print("=" * 60)
    print("[INFO] Custom Dictionary Loaded")
    print("=" * 60)
    print(f"Total Custom Words : {len(custom_words)}")

    print("\nCustom Words:")
    for word in custom_words:
        print(word)

    return custom_words


custom_words = load_custom_words("custom_words.csv")
```

```python
import jieba
import jieba.posseg as pseg
import re

# ================================
# Text Segmentation(新)
# ================================

# 保留的詞性
KEEP_POS = {
    "n",      # 一般名詞
    "nr",     # 人名
    "ns",     # 地名
    "nt",     # 組織名
    "nz",     # 其他專有名詞
    "eng"     # 英文
}

def segment_text(df, stopwords, custom_words=None):

    print("=" * 60)
    print("[INFO] Start Word Segmentation...")
    print("=" * 60)

    # 自訂詞集合
    if custom_words is None:
        custom_words = []

    custom_word_set = set(custom_words)

    token_list = []

    for text in df["clean_content"]:

        tokens = []

        for word, flag in pseg.cut(str(text)):

            word = word.strip()

            # 空字串
            if word == "":
                continue

            # 停用詞
            if word in stopwords:
                continue

            # 空白
            if word.isspace():
                continue

            # 純數字
            if word.isdigit():
                continue

            # 長度過短
            if len(word) <= 1:
                continue

            # 自訂詞優先保留
            if word in custom_word_set:
                tokens.append(word)
                continue

            # 詞性過濾
            if flag not in KEEP_POS:
                continue

            tokens.append(word)

        token_list.append(tokens)

    df["tokens"] = token_list

    print("[INFO] Word Segmentation Finished")
    print(f"[INFO] Total Posts : {len(df)}")
    print(f"[INFO] Custom Words : {len(custom_word_set)}")

    return df
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import numpy as np

# ==========================================
# TF-IDF Keyword Extraction
# ==========================================

def extract_keywords(df, top_n=5):

    print("=" * 60)
    print("[INFO] Start TF-IDF Keyword Extraction...")
    print("=" * 60)

    # 將 tokens 合併成字串
    documents = []

    for tokens in df["tokens"]:
      if len(tokens) == 0:
        documents.append("empty")
      else:
        documents.append(" ".join(tokens))

    vectorizer = TfidfVectorizer(
        tokenizer=lambda x: x.split(),
        lowercase=False,
        sublinear_tf=True,
        norm="l2"
    )

    tfidf_matrix = vectorizer.fit_transform(documents)

    feature_names = np.array(
        vectorizer.get_feature_names_out()
    )

    keyword_results = []

    for i in range(tfidf_matrix.shape[0]):

        row = tfidf_matrix[i].toarray().flatten()

        # 依 TF-IDF 分數排序
        top_index = row.argsort()[::-1][:top_n]

        keywords = []

        for idx in top_index:

            if row[idx] <= 0:
                continue

            keywords.append({
                "keyword": feature_names[idx],
                "score": round(float(row[idx]), 4)
            })

        keyword_results.append(keywords)

    df["keywords"] = keyword_results

    print("[INFO] Keyword Extraction Finished")

    return df
```

```python
# ==========================================
# Sentiment Analysis
# ==========================================

from transformers import pipeline

print("[INFO] Loading Sentiment Model...")

sentiment_pipe = pipeline(
    "text-classification",
    model="tabularisai/multilingual-sentiment-analysis",
    top_k=None
)

print("[INFO] Model Loaded")


def analyze_emotion(df):

    print("=" * 60)
    print("[INFO] Start Sentiment Analysis...")
    print("=" * 60)

    emotions = []
    emotion_scores = []
    emotion_details = []

    for text in df["clean_content"]:

        text = str(text)

        if text.strip() == "":

            emotions.append("Unknown")
            emotion_scores.append(0.0)
            emotion_details.append({})

            continue

        try:

            result = sentiment_pipe(text[:512])[0]

            detail = {}

            for item in result:
                detail[item["label"]] = round(
                    float(item["score"]), 4
                )

            best = max(
                result,
                key=lambda x: x["score"]
            )

            emotions.append(best["label"])

            emotion_scores.append(
                round(float(best["score"]), 4)
            )

            emotion_details.append(detail)

        except Exception:

            emotions.append("Error")
            emotion_scores.append(0.0)
            emotion_details.append({})

    df["emotion"] = emotions
    df["emotion_score"] = emotion_scores
    df["emotion_detail"] = emotion_details

    print("[INFO] Sentiment Analysis Finished")

    return df
```

```python
# ==========================================
# Daily Trend Analysis
# ==========================================

from collections import Counter


def daily_trend(df):

    print("=" * 60)
    print("Daily Trend Analysis")
    print("=" * 60)

    trend = {}

    dates = sorted(df["date"].unique())

    for date in dates:

        day_df = df[df["date"] == date]

        # 貼文數
        post_count = len(day_df)

        # 情緒統計
        emotion_count = (
            day_df["emotion"]
            .value_counts()
            .to_dict()
        )

        # 關鍵字統計
        keyword_counter = Counter()

        for kws in day_df["keywords"]:

            if kws is None:
                continue

            for item in kws:

                keyword_counter[item["keyword"]] += 1

        trend[date] = {
            "posts": post_count,
            "emotion": emotion_count,
            "top_keywords":
                keyword_counter.most_common(10)
        }

    return trend
```

```python
# ==========================================
# POS Filtering
# ==========================================

import jieba.posseg as pseg

# 保留的詞性
KEEP_POS = {
    "n",      # 一般名詞
    "nr",     # 人名
    "ns",     # 地名
    "nt",     # 組織
    "nz",     # 其他專有名詞
    "eng"     # 英文
}

def tokenize_with_pos(text, stopwords):

    words = []

    for word, flag in pseg.cut(text):

        word = word.strip()

        if len(word) <= 1:
            continue

        if word in stopwords:
            continue

        if flag not in KEEP_POS:
            continue

        words.append(word)

    return words
```

```python
# ==========================================
# Load Known Fake News Dataset
# ==========================================

import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity


def load_known_fake_news(filename):

    fake_df = pd.read_csv(filename)

    fake_df["title"] = fake_df["title"].fillna("")
    fake_df["content"] = fake_df["content"].fillna("")

    fake_df["text"] = (
        fake_df["title"].astype(str)
        + " "
        + fake_df["content"].astype(str)
    )

    return fake_df
```

```python
# ==========================================
# Fake News Similarity Detection
# ==========================================

def calculate_fake_similarity(df, fake_df):

    all_texts = (
        df["title"].fillna("").astype(str)
        + " "
        + df["content"].fillna("").astype(str)
    )

    fake_texts = fake_df["text"].tolist()

    if len(fake_texts) == 0:

        df["similarity_to_fake"] = 0.0

        return df

    vectorizer = TfidfVectorizer(
        analyzer="char",
        ngram_range=(2, 5),
        min_df=1
    )

    combined_texts = list(all_texts) + fake_texts

    tfidf_matrix = vectorizer.fit_transform(combined_texts)

    news_matrix = tfidf_matrix[:len(all_texts)]
    fake_matrix = tfidf_matrix[len(all_texts):]

    similarity_matrix = cosine_similarity(
        news_matrix,
        fake_matrix
    )

    max_similarity = similarity_matrix.max(axis=1)

    df["similarity_to_fake"] = max_similarity

    return df
```

```python
# ==========================================
# Suspicious Message Detection(新)
# ==========================================

import re

SUSPICIOUS_WORDS = [
    "快分享",
    "快轉",
    "立即分享",
    "不要刪",
    "一定要看",
    "震驚",
    "爆料",
    "獨家",
    "千萬不要",
    "一定是真的",
    "立即轉發",
    "分享出去"
]


def detect_suspicious(df, fake_df=None):

    print("=" * 60)
    print("[INFO] Detect Suspicious Messages...")
    print("=" * 60)

    if fake_df is not None:

        df = calculate_fake_similarity(
            df,
            fake_df
        )

    else:

        df["similarity_to_fake"] = 0.0

    scores = []
    levels = []
    reasons = []

    for _, row in df.iterrows():

        score = 0
        reason = []

        title = str(row["title"])
        content = str(row["content"])
        emotion = str(row["emotion"])

        # -----------------------
        # 情緒
        # -----------------------

        if emotion == "Very Negative":

            score += 2
            reason.append("Very Negative Emotion")

        elif emotion == "Negative":

            score += 1
            reason.append("Negative Emotion")

        # -----------------------
        # 煽動性文字
        # -----------------------

        for word in SUSPICIOUS_WORDS:

            if word in title or word in content:

                score += 2
                reason.append(
                    f"Keyword: {word}"
                )

        # -----------------------
        # 太多驚嘆號
        # -----------------------

        exclamation = (
            title.count("!")
            + content.count("!")
        )

        if exclamation >= 3:

            score += 1
            reason.append(
                "Many Exclamation Marks"
            )

        # -----------------------
        # 太多問號
        # -----------------------

        question = (
            title.count("?")
            + content.count("?")
        )

        if question >= 3:

            score += 1
            reason.append(
                "Many Question Marks"
            )

        # -----------------------
        # 全大寫英文
        # -----------------------

        if re.fullmatch(
            r"[A-Z0-9 ]{5,}",
            title
        ):

            score += 1
            reason.append(
                "Uppercase Title"
            )

        # -----------------------
        # 文章太長
        # -----------------------

        if len(content) > 300:

            score += 1
            reason.append(
                "Long Content"
            )

        # -----------------------
        # 已確認假訊息相似度
        # -----------------------

        similarity = float(
            row["similarity_to_fake"]
        )

        if similarity >= 0.85:

            score += 3

            reason.append(
                f"Highly Similar to Known Fake News ({similarity:.2f})"
            )

        elif similarity >= 0.70:

            score += 2

            reason.append(
                f"Similar to Known Fake News ({similarity:.2f})"
            )

        elif similarity >= 0.50:

            score += 1

            reason.append(
                f"Some Similarity to Known Fake News ({similarity:.2f})"
            )

        # -----------------------
        # 可疑等級
        # -----------------------

        if score >= 5:

            level = "High"

        elif score >= 3:

            level = "Medium"

        else:

            level = "Low"

        scores.append(score)
        levels.append(level)
        reasons.append(reason)

    df["suspicious_score"] = scores
    df["suspicious_level"] = levels
    df["suspicious_reason"] = reasons

    print(
        "[INFO] Suspicious Detection Finished"
    )

    return df
```

```python
# ==========================================
# Show Suspicious Result
# ==========================================

def show_suspicious(df, index):

    row = df.iloc[index]

    print("=" * 60)

    print("ID")
    print(row["id"])

    print()

    print("Title")
    print(row["title"])

    print()

    print("Level")
    print(row["suspicious_level"])

    print()

    print("Score")
    print(row["suspicious_score"])

    print()

    print("Reasons")

    if len(row["suspicious_reason"]) == 0:

        print("None")

    else:

        for r in row["suspicious_reason"]:

            print("-", r)
```

```python
# ==========================================
# Export Dictionary
# ==========================================

def export_dict(df):

    result = []

    for _, row in df.iterrows():

        item = {

            "id": row["id"],

            "platform": row["platform"],

            "title": row["title"],

            "content": row["content"],

            "date": row["date"],

            "keywords": row["keywords"],

            "emotion": {
                "label": row["emotion"],
                "score": round(float(row["emotion_score"]), 4)
            },

            "suspicious": {
                "score": row["suspicious_score"],
                "level": row["suspicious_level"],
                "reason": row["suspicious_reason"]
            }

        }

        result.append(item)

    return result
```
